```python
cut_order=['Ideal', 'Premium', 'Very Good', 'Good', 'Fair']
colour_order=['D', 'E', 'F', 'G', 'H', 'I', 'J']
clarity_order=['IF', 'VVS1', 'VVS2', 'VS1', 'VS2', 'SI1', 'SI2', 'I1']
```

```python
from sklearn.preprocessing import StandardScaler,OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=[cut_order,colour_order,clarity_order])
df[["cut","color","clarity"]]=oe.fit_transform(df[["cut","color","clarity"]])
```

```python
df.head()
```

```python
x=df.drop("cut",axis=1)
y=df["cut"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.20)
```

```python
sc=StandardScaler()
```

```python
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
knn=KNeighborsClassifier(n_neighbors=15)
```

```python
knn.fit(x_train_sc,y_train)
```

```python
knn.score(x_test_sc,y_test)
```

```python
knn_pred=knn.predict(x_test_sc)
knn_pred
```

```python
for k in range(3, 21, 2):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train_sc, y_train)
    print(f"k={k}  accuracy: {knn.score(x_test_sc, y_test):.4f}")
    
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
ml=RandomForestClassifier(n_estimators=700,random_state=42)
```

```python
ml.fit(x_train,y_train)
```

```python
ml.score(x_test,y_test)*100
```

```python
rl_pre=ml.predict(x_test)
rl_pre
```

```python
for n in [100, 200, 300, 500]:
    rf = RandomForestClassifier(n_estimators=n, random_state=42)
    rf.fit(x_train, y_train)
    print(f"n_estimators={n}  score: {rf.score(x_test, y_test):.4f}")
```

```python
# Create DataFrames
# -------------------
df_knn = pd.DataFrame({
    "Actual": y_test,
    "Predicted": knn_pred,
    "Model": "KNN"
})
df_rf = pd.DataFrame({
    "Actual": y_test,
    "Predicted": rl_pre,
    "Model": "Random Forest"
})

# Combine both
df_all = pd.concat([df_knn, df_rf])
plt.figure(figsize=(6, 4))

sns.scatterplot(
    data=df_all,
    x="Actual",
    y="Predicted",
    hue="Model",
    s=80
)

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color='red',
    linestyle='--'
)

plt.title("Actual vs Predicted")
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# ---------- KNN ----------
x_knn = y_test + np.random.normal(0, 0.05, len(y_test))
y_knn = knn_pred + np.random.normal(0, 0.05, len(knn_pred))

sns.scatterplot(
    x=x_knn,
    y=y_knn,
    ax=axes[0],
    color='blue',
    alpha=0.6,
    s=60
)
axes[0].plot(
    [0, 4],
    [0, 4],
    'r--',
    linewidth=2
)

axes[0].set_title("KNN: Actual vs Predicted")
axes[0].set_xlabel("Actual")
axes[0].set_ylabel("Predicted")

# ---------- Random Forest ----------
x_rf = y_test + np.random.normal(0, 0.05, len(y_test))
y_rf = rl_pre + np.random.normal(0, 0.05, len(rl_pre))

sns.scatterplot(
    x=x_rf,
    y=y_rf,
    ax=axes[1],
    color='orange',
    alpha=0.6,
    s=60
)
axes[1].plot(
    [0, 4],
    [0, 4],
    'r--',
    linewidth=2
)

axes[1].set_title("Random Forest: Actual vs Predicted")
axes[1].set_xlabel("Actual")
axes[1].set_ylabel("Predicted")

plt.tight_layout()
plt.show()
```
